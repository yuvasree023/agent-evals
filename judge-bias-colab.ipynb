# Separate frozen-judge bias experiment
48 uncached free Gemini verdict calls on 12 selected v2 dev traces. No Groq execution, holdout, labels or rubric changes. Stop and export partial evidence on quota/error. Not accuracy.

```
from pathlib import Path
import os,sys,subprocess,json,zipfile,base64
from google.colab import userdata
ROOT=Path('/content/agent-evals-bias')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/yuvasree023/agent-evals.git',str(ROOT)],check=True)
os.chdir(ROOT)
print('BIAS_SOURCE',subprocess.check_output(['git','rev-parse','HEAD'],text=True).strip())
sys.path.insert(0,str(ROOT))
os.environ['GEMINI_API_KEY']=userdata.get('GEMINI_API_KEY')
subprocess.run(['python3','-m','unittest','discover','-s','tests'],check=True)
from judge_bias_run import run
out=run()
z=Path('/content/judge-bias-results.zip')
with zipfile.ZipFile(z,'w',zipfile.ZIP_DEFLATED) as archive:
    for p in sorted(out.iterdir()):archive.write(p,p.name)
print('BIAS_RESULTS_BEGIN'+base64.b64encode(z.read_bytes()).decode()+'BIAS_RESULTS_END')
```
