# Judge validation (dev labels only)
Runs the Gemini Flash Lite judge on the 40 non-train labelled dev runs. Never touches the holdout. Add GEMINI_API_KEY in Colab Secrets and enable notebook access.

```
import subprocess, os
from pathlib import Path
ROOT=Path('/content/agent-evals')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/yuvasree023/agent-evals.git',str(ROOT)],check=True)
else:
    subprocess.run(['git','-C',str(ROOT),'pull','--quiet'],check=True)
from google.colab import userdata
os.environ['GEMINI_API_KEY']=userdata.get('GEMINI_API_KEY')
print('ok')
```

```
r=subprocess.run(['python3','judge/validate.py','results/judge-validation-dev-2026-10-12.json'],cwd=ROOT,capture_output=True,text=True)
print(r.stdout[-3000:]); print(r.stderr[-1500:])
```
